# Read Kode Wilayah Parquet File

This notebook demonstrates how to read the kode_wilayah.parquet file using Polars.

```python
import polars as pl

# Read the parquet file
df = pl.read_parquet('../output/parquet/provinsi.parquet')

# Display the dataframe
print("DataFrame shape:", df.shape)
print("Columns:", df.columns)
print(df)
```

```python
# Additional analysis
print("Data types:")
print(df.dtypes)

print("\nSummary statistics:")
print(df.describe())
```
